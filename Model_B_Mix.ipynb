```python
from PIL import Image, ImageOps
from torch.utils.data import Dataset

class ImagePathDataset(Dataset):

    def __init__(
        self,
        samples,
        indices,
        transform
    ):
        self.samples = [
            samples[index]
            for index in indices
        ]

        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        image_path, label = self.samples[index]

        with Image.open(image_path) as image:
            image = ImageOps.exif_transpose(
                image
            ).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        return image, label
```

```python
import torch
import torch.nn as nn

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.0001,
    weight_decay=0.0001
)

use_mixed_precision = device.type == "cuda"

scaler = torch.amp.GradScaler(
    "cuda",
    enabled=use_mixed_precision
)
```

```python
from tqdm.auto import tqdm

model.train()

running_loss = 0.0
correct_predictions = 0
images_processed = 0

progress_bar = tqdm(
    train_loader,
    desc="Training epoch 1"
)

for images, labels in progress_bar:

    images = images.to(
        device,
        non_blocking=True
    )

    labels = labels.to(
        device,
        non_blocking=True
    )

    optimizer.zero_grad(
        set_to_none=True
    )

    with torch.autocast(
        device_type=device.type,
        dtype=torch.float16,
        enabled=use_mixed_precision
    ):
        outputs = model(images)
        loss = criterion(outputs, labels)

    scaler.scale(loss).backward()
    scaler.step(optimizer)
    scaler.update()

    batch_size = images.size(0)

    running_loss += (
        loss.item() * batch_size
    )

    predictions = outputs.argmax(
        dim=1
    )

    correct_predictions += (
        predictions == labels
    ).sum().item()

    images_processed += batch_size

    current_loss = (
        running_loss / images_processed
    )

    current_accuracy = (
        correct_predictions
        / images_processed
    )

    progress_bar.set_postfix(
        loss=f"{current_loss:.4f}",
        accuracy=f"{current_accuracy:.4f}"
    )

training_loss = (
    running_loss / images_processed
)

training_accuracy = (
    correct_predictions / images_processed
)

print("Epoch 1 completed")
print("Training loss:", training_loss)
print("Training accuracy:", training_accuracy)
```

```python
from torch.utils.data import DataLoader
from sklearn.metrics import roc_auc_score
from tqdm.auto import tqdm
import torch

def evaluate_dataset(
    dataset,
    description,
    batch_size=64
):
    loader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    model.eval()

    all_labels = []
    all_ai_scores = []

    correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in tqdm(
            loader,
            desc=description
        ):

            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            with torch.autocast(
                device_type=device.type,
                dtype=torch.float16,
                enabled=use_mixed_precision
            ):
                outputs = model(images)

            probabilities = torch.softmax(
                outputs,
                dim=1
            )

            ai_probabilities = probabilities[
                :,
                AI_CLASS_INDEX
            ]

            predictions = outputs.argmax(
                dim=1
            )

            correct += (
                predictions == labels
            ).sum().item()

            total += labels.size(0)

            binary_ai_labels = (
                labels == AI_CLASS_INDEX
            ).int()

            all_labels.extend(
                binary_ai_labels
                .cpu()
                .numpy()
                .tolist()
            )

            all_ai_scores.extend(
                ai_probabilities
                .cpu()
                .numpy()
                .tolist()
            )

    accuracy = correct / total

    auc = roc_auc_score(
        all_labels,
        all_ai_scores
    )

    return {
        "images": total,
        "accuracy": accuracy,
        "auc": auc
    }
```

```python
import io
import numpy as np

from PIL import (
    Image,
    ImageEnhance,
    ImageFilter
)

from torchvision import transforms
```

```python
class JPEGCompression:

    def __init__(self, quality):
        self.quality = quality

    def __call__(self, image):

        buffer = io.BytesIO()

        image.save(
            buffer,
            format="JPEG",
            quality=self.quality
        )

        buffer.seek(0)

        with Image.open(buffer) as compressed:
            return compressed.convert("RGB")
```

```python
class DownscaleUpscale:

    def __init__(self, scale):
        self.scale = scale

    def __call__(self, image):

        original_width, original_height = (
            image.size
        )

        smaller_size = (
            max(
                1,
                round(
                    original_width
                    * self.scale
                )
            ),
            max(
                1,
                round(
                    original_height
                    * self.scale
                )
            )
        )

        smaller = image.resize(
            smaller_size,
            Image.Resampling.BICUBIC
        )

        return smaller.resize(
            (
                original_width,
                original_height
            ),
            Image.Resampling.BICUBIC
        )
```

```python
class DownscaleUpscale:

    def __init__(self, scale):
        self.scale = scale

    def __call__(self, image):

        original_width, original_height = (
            image.size
        )

        smaller_size = (
            max(
                1,
                round(
                    original_width
                    * self.scale
                )
            ),
            max(
                1,
                round(
                    original_height
                    * self.scale
                )
            )
        )

        smaller = image.resize(
            smaller_size,
            Image.Resampling.BICUBIC
        )

        return smaller.resize(
            (
                original_width,
                original_height
            ),
            Image.Resampling.BICUBIC
        )
```

```python
class CenterCropPercent:

    def __init__(self, percentage):
        self.percentage = percentage

    def __call__(self, image):

        width, height = image.size

        crop_width = round(
            width * self.percentage
        )

        crop_height = round(
            height * self.percentage
        )

        left = (
            width - crop_width
        ) // 2

        top = (
            height - crop_height
        ) // 2

        cropped = image.crop(
            (
                left,
                top,
                left + crop_width,
                top + crop_height
            )
        )

        return cropped.resize(
            (width, height),
            Image.Resampling.BICUBIC
        )
```

```python
class FixedColorAdjustment:

    def __init__(
        self,
        brightness,
        contrast,
        saturation
    ):
        self.brightness = brightness
        self.contrast = contrast
        self.saturation = saturation

    def __call__(self, image):

        image = ImageEnhance.Brightness(
            image
        ).enhance(
            self.brightness
        )

        image = ImageEnhance.Contrast(
            image
        ).enhance(
            self.contrast
        )

        image = ImageEnhance.Color(
            image
        ).enhance(
            self.saturation
        )

        return image
```

```python
class AddGaussianNoise:

    def __init__(self, sigma):
        self.sigma = sigma

    def __call__(self, tensor):

        noise = torch.randn_like(
            tensor
        ) * self.sigma

        return torch.clamp(
            tensor + noise,
            0.0,
            1.0
        )
```

```python
def evaluate_model(dataset, name, batch_size=64):
    loader = DataLoader(
        dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    model.eval()

    evaluation_labels = []
    evaluation_probabilities = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)

            outputs = model(images)
            probabilities = torch.softmax(outputs, dim=1)
            ai_probabilities = probabilities[:, AI_CLASS_INDEX]

            evaluation_labels.extend(labels.cpu().numpy())
            evaluation_probabilities.extend(
                ai_probabilities.cpu().numpy()
            )

    evaluation_labels = np.array(evaluation_labels)
    evaluation_probabilities = np.array(evaluation_probabilities)

    binary_labels = (
        evaluation_labels == AI_CLASS_INDEX
    ).astype(int)

    predictions = (
        evaluation_probabilities >= 0.5
    ).astype(int)

    accuracy = accuracy_score(binary_labels, predictions)
    roc_auc = roc_auc_score(
        binary_labels,
        evaluation_probabilities
    )

    result = {
        "condition": name,
        "images": len(binary_labels),
        "accuracy": float(accuracy),
        "roc_auc": float(roc_auc)
    }

    print(name)
    print("Images:", result["images"])
    print("Accuracy:", round(result["accuracy"], 4))
    print("ROC AUC:", round(result["roc_auc"], 4))

    return result
```

```python
from torchvision.transforms import functional as TF
from torchvision.transforms import InterpolationMode


class DownscaleUpscale:
    def __init__(self, scale):
        self.scale = scale

    def __call__(self, image):
        original_width, original_height = image.size

        small_width = max(1, int(original_width * self.scale))
        small_height = max(1, int(original_height * self.scale))

        image = TF.resize(
            image,
            [small_height, small_width],
            interpolation=InterpolationMode.BILINEAR
        )

        image = TF.resize(
            image,
            [original_height, original_width],
            interpolation=InterpolationMode.BILINEAR
        )

        return image
```

```python
class AddGaussianNoise:
    def __init__(self, sigma):
        self.sigma = sigma

    def __call__(self, tensor):
        noise = torch.randn_like(tensor) * self.sigma
        noisy_tensor = tensor + noise

        return torch.clamp(
            noisy_tensor,
            min=0.0,
            max=1.0
        )
```

```python
import random


class RandomGaussianNoise:
    def __init__(
        self,
        probability=0.5,
        sigma_choices=(0.02, 0.05, 0.10)
    ):
        self.probability = probability
        self.sigma_choices = sigma_choices

    def __call__(self, tensor):
        if random.random() >= self.probability:
            return tensor

        sigma = random.choice(self.sigma_choices)
        noise = torch.randn_like(tensor) * sigma

        return torch.clamp(
            tensor + noise,
            min=0.0,
            max=1.0
        )
```

```python
import random


class RandomGaussianNoise:
    def __init__(
        self,
        probability=0.5,
        sigma_choices=(0.02, 0.05, 0.10)
    ):
        self.probability = probability
        self.sigma_choices = sigma_choices

    def __call__(self, tensor):
        if random.random() >= self.probability:
            return tensor

        sigma = random.choice(self.sigma_choices)
        noise = torch.randn_like(tensor) * sigma

        return torch.clamp(
            tensor + noise,
            min=0.0,
            max=1.0
        )
```

```python
sid_train_transform = transforms.Compose([
    # Randomly crop the high-resolution image and output 224×224
    transforms.RandomResizedCrop(
        224,
        scale=(0.8, 1.0)
    ),

    # Sometimes mirror the image
    transforms.RandomHorizontalFlip(p=0.5),

    # Simulate colour changes
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2
    ),

    # Sometimes simulate blur
    transforms.RandomApply(
        [
            transforms.GaussianBlur(
                kernel_size=5,
                sigma=(0.5, 2.0)
            )
        ],
        p=0.3
    ),

    # Convert the image into model-readable numbers
    transforms.ToTensor(),

    # Sometimes add one of the official noise levels
    RandomGaussianNoise(
        probability=0.5,
        sigma_choices=(0.02, 0.05, 0.10)
    ),

    # Apply the normal ResNet50 input normalisation
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
from torch.utils.data import Dataset, DataLoader


REAL_CLASS_INDEX = checkpoint["class_to_idx"]["REAL"]


class SIDTrainingDataset(Dataset):
    def __init__(self, huggingface_dataset, transform):
        self.dataset = huggingface_dataset
        self.transform = transform

        self.indices = [
            index
            for index, label
            in enumerate(huggingface_dataset["label"])
            if label in [0, 1]
        ]

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, position):
        dataset_index = self.indices[position]
        example = self.dataset[dataset_index]

        image = example["image"].convert("RGB")
        sid_label = example["label"]

        if sid_label == 1:
            model_label = AI_CLASS_INDEX
        else:
            model_label = REAL_CLASS_INDEX

        image = self.transform(image)

        return image, model_label


sid_training_dataset = SIDTrainingDataset(
    sid_train,
    sid_train_transform
)

sid_training_loader = DataLoader(
    sid_training_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

print("Model B-HR training images:", len(sid_training_dataset))
print("AI output index:", AI_CLASS_INDEX)
print("Real output index:", REAL_CLASS_INDEX)
```

```python
# Freeze the entire model
for parameter in model.parameters():
    parameter.requires_grad = False

# Allow the final ResNet block to adapt
for parameter in model.layer4.parameters():
    parameter.requires_grad = True

# Allow the classifier to adapt
for parameter in model.fc.parameters():
    parameter.requires_grad = True

trainable_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)

total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print("Trainable parameters:", trainable_parameters)
print("Total parameters:", total_parameters)
print(
    "Trainable percentage:",
    round(100 * trainable_parameters / total_parameters, 2),
    "%"
)
```

```python
import torch.nn as nn

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.AdamW(
    [
        {
            "params": model.layer4.parameters(),
            "lr": 0.00001
        },
        {
            "params": model.fc.parameters(),
            "lr": 0.00005
        }
    ],
    weight_decay=0.0001
)
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score


def evaluate_sid_model(model, loader):
    model.eval()

    evaluation_labels = []
    evaluation_probabilities = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)

            outputs = model(images)
            probabilities = torch.softmax(
                outputs,
                dim=1
            )

            ai_probabilities = probabilities[
                :,
                AI_CLASS_INDEX
            ]

            evaluation_labels.extend(
                labels.numpy()
            )

            evaluation_probabilities.extend(
                ai_probabilities.cpu().numpy()
            )

    evaluation_labels = np.array(evaluation_labels)
    evaluation_probabilities = np.array(
        evaluation_probabilities
    )

    predictions = (
        evaluation_probabilities >= 0.5
    ).astype(int)

    return {
        "accuracy": float(
            accuracy_score(
                evaluation_labels,
                predictions
            )
        ),
        "roc_auc": float(
            roc_auc_score(
                evaluation_labels,
                evaluation_probabilities
            )
        )
    }
```

```python
from tqdm.auto import tqdm

best_sid_auc = 0.0
best_epoch = 0

training_history = []

for epoch in range(1, 4):
    # Keep frozen sections in evaluation mode.
    # Train only layer4 and the classifier.
    model.eval()
    model.layer4.train()
    model.fc.train()

    running_loss = 0.0
    correct_predictions = 0
    total_images = 0

    progress_bar = tqdm(
        sid_training_loader,
        desc=f"Model B-HR epoch {epoch}/3"
    )

    for images, labels in progress_bar:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad(set_to_none=True)

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += (
            loss.item() * images.size(0)
        )

        predictions = outputs.argmax(dim=1)

        correct_predictions += (
            predictions == labels
        ).sum().item()

        total_images += labels.size(0)

        progress_bar.set_postfix(
            loss=round(loss.item(), 4)
        )

    training_loss = running_loss / total_images
    training_accuracy = (
        correct_predictions / total_images
    )

    validation_result = evaluate_sid_model(
        model,
        sid_loader
    )

    epoch_result = {
        "epoch": epoch,
        "training_loss": training_loss,
        "training_accuracy": training_accuracy,
        "sid_validation_accuracy": (
            validation_result["accuracy"]
        ),
        "sid_validation_auc": (
            validation_result["roc_auc"]
        )
    }

    training_history.append(epoch_result)

    print()
    print("Epoch:", epoch)
    print("Training loss:", round(training_loss, 4))
    print(
        "Training accuracy:",
        round(training_accuracy, 4)
    )
    print(
        "SID validation accuracy:",
        round(validation_result["accuracy"], 4)
    )
    print(
        "SID validation ROC AUC:",
        round(validation_result["roc_auc"], 4)
    )

    if validation_result["roc_auc"] > best_sid_auc:
        best_sid_auc = validation_result["roc_auc"]
        best_epoch = epoch

        torch.save(
            {
                "model_state_dict": model.state_dict(),
                "class_to_idx": checkpoint["class_to_idx"],
                "ai_class_index": AI_CLASS_INDEX,
                "experiment": "Model B-HR",
                "best_epoch": best_epoch,
                "sid_validation_auc": best_sid_auc,
                "training_history": training_history
            },
            "/content/resnet50_model_b_hr_best.pth"
        )

        print("New best Model B-HR saved.")

    print("-" * 50)

print("Best epoch:", best_epoch)
print("Best SID validation AUC:", round(best_sid_auc, 4))
```

```python
import numpy as np
import torch

from torch.utils.data import DataLoader
from sklearn.metrics import accuracy_score, roc_auc_score


def evaluate_cifake(model, dataset, name):
    loader = DataLoader(
        dataset,
        batch_size=64,
        shuffle=False,
        num_workers=0,
        pin_memory=torch.cuda.is_available()
    )

    model.eval()

    labels_list = []
    probabilities_list = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)

            outputs = model(images)

            probabilities = torch.softmax(
                outputs,
                dim=1
            )

            ai_probabilities = probabilities[
                :,
                AI_CLASS_INDEX
            ]

            labels_list.extend(
                labels.cpu().numpy()
            )

            probabilities_list.extend(
                ai_probabilities.cpu().numpy()
            )

    labels_array = np.array(labels_list)
    probabilities_array = np.array(
        probabilities_list
    )

    binary_ai_labels = (
        labels_array == AI_CLASS_INDEX
    ).astype(int)

    predicted_ai_labels = (
        probabilities_array >= 0.5
    ).astype(int)

    result = {
        "accuracy": float(
            accuracy_score(
                binary_ai_labels,
                predicted_ai_labels
            )
        ),
        "roc_auc": float(
            roc_auc_score(
                binary_ai_labels,
                probabilities_array
            )
        )
    }

    print(name)
    print("Images:", len(labels_array))
    print("Accuracy:", round(result["accuracy"], 4))
    print("ROC AUC:", round(result["roc_auc"], 4))

    return result
```

```python
from pathlib import Path
from PIL import Image

import numpy as np
import torch

from torchvision import datasets, transforms
from torch.utils.data import Dataset
from sklearn.model_selection import train_test_split


# 1. Confirm CIFAKE folders exist

train_path = Path("/content/training/train")
test_path = Path("/content/training/test")

assert train_path.exists(), f"Missing: {train_path}"
assert test_path.exists(), f"Missing: {test_path}"

print("CIFAKE folders found.")


# 2. Read the original archive folders

original_train_dataset = datasets.ImageFolder(
    train_path
)

original_test_dataset = datasets.ImageFolder(
    test_path
)

assert (
    original_train_dataset.class_to_idx
    == original_test_dataset.class_to_idx
)

print(
    "Class mapping:",
    original_train_dataset.class_to_idx
)


# 3. Combine the original train and test samples

all_samples = (
    original_train_dataset.samples
    + original_test_dataset.samples
)

source_labels = np.array([
    label for _, label in all_samples
])

all_indices = np.arange(len(all_samples))


# 4. Recreate the exact 70/15/15 split

train_indices, remaining_indices = train_test_split(
    all_indices,
    test_size=0.30,
    random_state=2026,
    stratify=source_labels
)

validation_indices, test_indices = train_test_split(
    remaining_indices,
    test_size=0.50,
    random_state=2026,
    stratify=source_labels[remaining_indices]
)

print("Training images:", len(train_indices))
print("Validation images:", len(validation_indices))
print("Test images:", len(test_indices))


# 5. Define the image-loading dataset

class CIFAKEPathDataset(Dataset):
    def __init__(self, samples, indices, transform):
        self.samples = samples
        self.indices = indices
        self.transform = transform

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, position):
        sample_index = self.indices[position]
        image_path, label = self.samples[sample_index]

        image = Image.open(image_path).convert("RGB")
        image = self.transform(image)

        return image, label


# 6. Define clean ResNet preprocessing

cifake_clean_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# 7. Create the missing validation dataset

cifake_validation_dataset = CIFAKEPathDataset(
    all_samples,
    validation_indices,
    cifake_clean_transform
)

print(
    "CIFAKE validation dataset created:",
    len(cifake_validation_dataset)
)
```

```python
mixed_optimizer = torch.optim.AdamW(
    [
        {
            "params": mixed_model.layer4.parameters(),
            "lr": 0.00001
        },
        {
            "params": mixed_model.fc.parameters(),
            "lr": 0.00005
        }
    ],
    weight_decay=0.0001
)

mixed_history = []
best_mixed_score = 0.0
best_mixed_epoch = 0

for epoch in range(1, 4):
    mixed_model.eval()
    mixed_model.layer4.train()
    mixed_model.fc.train()

    running_loss = 0.0
    correct = 0
    total = 0

    progress_bar = tqdm(
        mixed_training_loader,
        desc=f"Model B-Mix epoch {epoch}/3"
    )

    for images, labels in progress_bar:
        images = images.to(device)
        labels = labels.to(device)

        mixed_optimizer.zero_grad(set_to_none=True)

        outputs = mixed_model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        mixed_optimizer.step()

        running_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)
        correct += (
            predictions == labels
        ).sum().item()

        total += labels.size(0)

        progress_bar.set_postfix(
            loss=round(loss.item(), 4)
        )

    training_loss = running_loss / total
    training_accuracy = correct / total

    sid_result = evaluate_sid_model(
        mixed_model,
        sid_loader
    )

    cifake_result = evaluate_cifake(
        mixed_model,
        cifake_validation_dataset,
        f"Epoch {epoch} — CIFAKE validation"
    )

    # Select a model that performs well on both sources
    selection_score = (
        sid_result["roc_auc"]
        + cifake_result["roc_auc"]
    ) / 2

    epoch_result = {
        "epoch": epoch,
        "training_loss": training_loss,
        "training_accuracy": training_accuracy,
        "sid_auc": sid_result["roc_auc"],
        "cifake_auc": cifake_result["roc_auc"],
        "mean_auc": selection_score
    }

    mixed_history.append(epoch_result)

    print("Epoch:", epoch)
    print("Training loss:", round(training_loss, 4))
    print("Training accuracy:", round(training_accuracy, 4))
    print("SID AUC:", round(sid_result["roc_auc"], 4))
    print("CIFAKE AUC:", round(cifake_result["roc_auc"], 4))
    print("Mean AUC:", round(selection_score, 4))

    if selection_score > best_mixed_score:
        best_mixed_score = selection_score
        best_mixed_epoch = epoch

        torch.save(
            {
                "model_state_dict": mixed_model.state_dict(),
                "class_to_idx": checkpoint["class_to_idx"],
                "ai_class_index": AI_CLASS_INDEX,
                "experiment": "Model B-Mix",
                "best_epoch": best_mixed_epoch,
                "selection_score": best_mixed_score,
                "training_history": mixed_history
            },
            "/content/resnet50_model_b_mix_best.pth"
        )

        print("New best Model B-Mix saved.")

    print("-" * 50)

print("Best mixed epoch:", best_mixed_epoch)
print("Best mean AUC:", round(best_mixed_score, 4))
```

```python
class FixedGaussianNoise:
    def __init__(self, sigma):
        self.sigma = sigma

    def __call__(self, tensor):
        noise = torch.randn_like(tensor) * self.sigma

        return torch.clamp(
            tensor + noise,
            min=0.0,
            max=1.0
        )
```

```python
import io
from PIL import Image


class JPEGCompression:
    def __init__(self, quality):
        self.quality = quality

    def __call__(self, image):
        buffer = io.BytesIO()

        image.save(
            buffer,
            format="JPEG",
            quality=self.quality
        )

        buffer.seek(0)

        compressed_image = Image.open(
            buffer
        ).convert("RGB")

        compressed_image.load()

        return compressed_image
```

```python
from torchvision.transforms import functional as TF
from torchvision.transforms import InterpolationMode


class DownscaleUpscale:
    def __init__(self, scale):
        self.scale = scale

    def __call__(self, image):
        width, height = image.size

        small_width = max(
            1,
            int(width * self.scale)
        )

        small_height = max(
            1,
            int(height * self.scale)
        )

        image = TF.resize(
            image,
            [small_height, small_width],
            interpolation=InterpolationMode.BILINEAR
        )

        image = TF.resize(
            image,
            [height, width],
            interpolation=InterpolationMode.BILINEAR
        )

        return image
```

```python
class FixedColorAdjustment:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, image):
        image = TF.adjust_brightness(
            image,
            self.factor
        )

        image = TF.adjust_contrast(
            image,
            self.factor
        )

        image = TF.adjust_saturation(
            image,
            self.factor
        )

        return image
```

```python
from pathlib import Path
from PIL import Image

import torch
import numpy as np

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms


test_root = Path("/content/manual_test")
real_folder = test_root / "Real Image"
ai_folder = test_root / "AI Image"

image_extensions = {
    ".jpg", ".jpeg", ".png",
    ".webp", ".bmp"
}

real_paths = [
    path
    for path in real_folder.rglob("*")
    if path.is_file()
    and path.suffix.lower() in image_extensions
]

ai_paths = [
    path
    for path in ai_folder.rglob("*")
    if path.is_file()
    and path.suffix.lower() in image_extensions
]

print("Real images:", len(real_paths))
print("AI images:", len(ai_paths))

assert len(real_paths) > 0, "No real images found."
assert len(ai_paths) > 0, "No AI images found."
```

```python
manual_test_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


class ManualImageDataset(Dataset):
    def __init__(
        self,
        real_paths,
        ai_paths,
        transform
    ):
        # Evaluation convention:
        # 0 = real
        # 1 = AI
        self.samples = (
            [(path, 0) for path in real_paths]
            + [(path, 1) for path in ai_paths]
        )

        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):
        image_path, label = self.samples[index]

        image = Image.open(
            image_path
        ).convert("RGB")

        image = self.transform(image)

        return image, label, str(image_path)


manual_test_dataset = ManualImageDataset(
    real_paths,
    ai_paths,
    manual_test_transform
)

manual_test_loader = DataLoader(
    manual_test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

print(
    "Total test images:",
    len(manual_test_dataset)
)
```

```python
from pathlib import Path
from PIL import Image
import torch
from torchvision import transforms
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix
import pandas as pd

# Confirm that the model is loaded
assert "mixed_model" in globals(), (
    "mixed_model is not loaded. Restore Model B-Mix before running this cell."
)

mixed_model = mixed_model.to(device)
mixed_model.eval()

# Same preprocessing used during validation
manual_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

real_folder = Path("/content/manual_test/Real Image")
ai_folder = Path("/content/manual_test/AI Image")

valid_extensions = {".jpg", ".jpeg", ".png", ".webp", ".bmp"}

test_files = []

for path in real_folder.iterdir():
    if path.suffix.lower() in valid_extensions:
        test_files.append((path, 0))  # 0 = Real

for path in ai_folder.iterdir():
    if path.suffix.lower() in valid_extensions:
        test_files.append((path, 1))  # 1 = AI

results = []

with torch.no_grad():
    for image_path, true_label in test_files:
        image = Image.open(image_path).convert("RGB")
        image_tensor = manual_transform(image).unsqueeze(0).to(device)

        outputs = mixed_model(image_tensor)
        probabilities = torch.softmax(outputs, dim=1)

        # Model output mapping: FAKE/AI = index 0
        ai_probability = probabilities[0, 0].item()
        real_probability = probabilities[0, 1].item()

        # Manual evaluation mapping: 0 = Real, 1 = AI
        predicted_label = 1 if ai_probability >= 0.5 else 0

        results.append({
            "filename": image_path.name,
            "actual": "AI" if true_label == 1 else "Real",
            "prediction": "AI" if predicted_label == 1 else "Real",
            "AI probability": ai_probability,
            "Real probability": real_probability,
            "correct": predicted_label == true_label,
            "true_label": true_label,
            "predicted_label": predicted_label,
        })

results_df = pd.DataFrame(results)

display(
    results_df[
        [
            "filename",
            "actual",
            "prediction",
            "AI probability",
            "Real probability",
            "correct",
        ]
    ].sort_values(["actual", "filename"])
)

true_labels = results_df["true_label"].tolist()
predicted_labels = results_df["predicted_label"].tolist()
ai_probabilities = results_df["AI probability"].tolist()

print("\nManual test accuracy:", round(
    accuracy_score(true_labels, predicted_labels), 4
))

print("Manual test ROC AUC:", round(
    roc_auc_score(true_labels, ai_probabilities), 4
))

print("\nConfusion matrix:")
print(confusion_matrix(true_labels, predicted_labels))
print("\nMatrix format: [[real predicted real, real predicted AI],")
print("                [AI predicted real, AI predicted AI]]")
```

```python
from pathlib import Path
from PIL import Image, ImageFilter, ImageEnhance
from io import BytesIO

import numpy as np
import pandas as pd
import torch

from torchvision import transforms
from sklearn.metrics import accuracy_score, roc_auc_score

assert "mixed_model" in globals(), "mixed_model is not loaded."
assert "device" in globals(), "device is not defined."

mixed_model = mixed_model.to(device)
mixed_model.eval()

# Model preprocessing
model_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

RESAMPLE = Image.Resampling.LANCZOS

def resize_224(image):
    return image.resize((224, 224), RESAMPLE)


def jpeg_compress(image, quality=30):
    buffer = BytesIO()

    image.save(
        buffer,
        format="JPEG",
        quality=quality
    )

    buffer.seek(0)

    return Image.open(buffer).convert("RGB")


def downscale_upscale(image, scale=0.5):
    width, height = image.size

    small_width = max(1, int(width * scale))
    small_height = max(1, int(height * scale))

    small = image.resize(
        (small_width, small_height),
        RESAMPLE
    )

    return small.resize(
        (width, height),
        RESAMPLE
    )


def centre_crop(image, retained_fraction=0.8):
    width, height = image.size

    crop_width = int(width * retained_fraction)
    crop_height = int(height * retained_fraction)

    left = (width - crop_width) // 2
    top = (height - crop_height) // 2

    cropped = image.crop((
        left,
        top,
        left + crop_width,
        top + crop_height
    ))

    return cropped.resize(
        (width, height),
        RESAMPLE
    )


def create_transformed_views(original_image):
    # First standardize every image to 224×224.
    image = resize_224(original_image.convert("RGB"))

    return {
        "original": image,
        "jpeg_q30": jpeg_compress(image, quality=30),
        "blur_1": image.filter(
            ImageFilter.GaussianBlur(radius=1.0)
        ),
        "resize_05": downscale_upscale(image, scale=0.5),
        "crop_80": centre_crop(image, retained_fraction=0.8),
        "colour_minus20": ImageEnhance.Color(image).enhance(0.8),
        "colour_plus20": ImageEnhance.Color(image).enhance(1.2),
    }


def predict_ai_probability(image):
    tensor = model_transform(image).unsqueeze(0).to(device)

    with torch.no_grad():
        output = mixed_model(tensor)
        probability = torch.softmax(output, dim=1)

    # Model training mapping:
    # output 0 = FAKE/AI
    # output 1 = REAL
    return probability[0, 0].item()


# Locate manual test images
real_folder = Path("/content/manual_test/Real Image")
ai_folder = Path("/content/manual_test/AI Image")

valid_extensions = {
    ".jpg", ".jpeg", ".png", ".webp", ".bmp"
}

manual_files = []

for path in real_folder.iterdir():
    if path.suffix.lower() in valid_extensions:
        manual_files.append((path, 0))  # 0 = real

for path in ai_folder.iterdir():
    if path.suffix.lower() in valid_extensions:
        manual_files.append((path, 1))  # 1 = AI


# Run every image through every transformed view
consensus_results = []

for image_path, true_label in manual_files:
    image = Image.open(image_path).convert("RGB")
    views = create_transformed_views(image)

    view_probabilities = {
        view_name: predict_ai_probability(view_image)
        for view_name, view_image in views.items()
    }

    probability_values = list(view_probabilities.values())

    mean_probability = float(
        np.mean(probability_values)
    )

    median_probability = float(
        np.median(probability_values)
    )

    probability_std = float(
        np.std(probability_values)
    )

    median_prediction = int(
        median_probability >= 0.5
    )

    view_predictions = [
        int(probability >= 0.5)
        for probability in probability_values
    ]

    consensus_agreement = max(
        view_predictions.count(0),
        view_predictions.count(1)
    ) / len(view_predictions)

    row = {
        "filename": image_path.name,
        "actual": "AI" if true_label == 1 else "Real",
        "true_label": true_label,
        **view_probabilities,
        "mean_AI_probability": mean_probability,
        "median_AI_probability": median_probability,
        "probability_std": probability_std,
        "agreement": consensus_agreement,
        "consensus_prediction": (
            "AI" if median_prediction == 1 else "Real"
        ),
        "consensus_correct": median_prediction == true_label,
        "consensus_label": median_prediction,
    }

    consensus_results.append(row)


consensus_df = pd.DataFrame(consensus_results)

display_columns = [
    "filename",
    "actual",
    "original",
    "jpeg_q30",
    "blur_1",
    "resize_05",
    "crop_80",
    "colour_minus20",
    "colour_plus20",
    "median_AI_probability",
    "probability_std",
    "agreement",
    "consensus_prediction",
    "consensus_correct",
]

display(
    consensus_df[display_columns]
    .sort_values(["actual", "filename"])
    .round(4)
)


# Compare original prediction against consensus prediction
true_labels = consensus_df["true_label"].tolist()

original_probabilities = consensus_df["original"].tolist()
mean_probabilities = consensus_df["mean_AI_probability"].tolist()
median_probabilities = consensus_df["median_AI_probability"].tolist()

original_predictions = [
    int(probability >= 0.5)
    for probability in original_probabilities
]

mean_predictions = [
    int(probability >= 0.5)
    for probability in mean_probabilities
]

median_predictions = consensus_df[
    "consensus_label"
].tolist()

print("\n===== Original single-view model =====")
print(
    "Accuracy:",
    round(
        accuracy_score(
            true_labels,
            original_predictions
        ),
        4
    )
)
print(
    "ROC AUC:",
    round(
        roc_auc_score(
            true_labels,
            original_probabilities
        ),
        4
    )
)

print("\n===== Mean transformation consensus =====")
print(
    "Accuracy:",
    round(
        accuracy_score(
            true_labels,
            mean_predictions
        ),
        4
    )
)
print(
    "ROC AUC:",
    round(
        roc_auc_score(
            true_labels,
            mean_probabilities
        ),
        4
    )
)

print("\n===== Median transformation consensus =====")
print(
    "Accuracy:",
    round(
        accuracy_score(
            true_labels,
            median_predictions
        ),
        4
    )
)
print(
    "ROC AUC:",
    round(
        roc_auc_score(
            true_labels,
            median_probabilities
        ),
        4
    )
)

print("\nMean transformation agreement:")
print(
    round(
        consensus_df["agreement"].mean(),
        4
    )
)
```
